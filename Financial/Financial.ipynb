```python
RD = df.groupby('Company')['R&D_Spending_USD_Mn'].sum()/1000

RD
```

```python
RV = df.groupby('Company')['AI_Revenue_USD_Mn'].sum()
RV
```

```python
plt.bar(RV.index, RV.values, color = ['navy', 'darkred', 'green'])
plt.title('AI Spend')
plt.xlabel('Company')
plt.ylabel('Amount')
```

```python
plt.figure(figsize = (12,6))

plt.subplot(1,2,1)

plt.bar(RD.index, RD.values, color = ['navy', 'darkred', 'green'])
plt.title('AI Spend')
plt.xlabel('Company')
plt.ylabel('Amount')

plt.subplot(1,2,2)

plt.bar(RV.index, RV.values, color = ['navy', 'darkred', 'green'])
plt.title('AI Revenue')
plt.xlabel('Company')
plt.ylabel('Revenue')
```
