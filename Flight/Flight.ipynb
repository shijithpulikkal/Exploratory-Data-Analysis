```python
plt.figure(figsize = (20,6))

plt.subplot(1,3,1)

plt.bar(df.departure_time.value_counts().index, df.departure_time.value_counts(ascending = False).values, color = ['red','grey'] )
plt.title('Departure_time')
plt.xlabel('time of Day')
plt.ylabel('number of flights')

plt.subplot(1,3,2)

plt.bar(df.arrival_time.value_counts().index, df.arrival_time.value_counts(ascending = False).values, color = ['green','yellow'] )
plt.title('Arrival_time')
plt.xlabel('time of Day')
plt.ylabel('number of flights')

plt.subplot(1,3,3)

plt.bar(df.arrival_time.value_counts().index, df.arrival_time.value_counts(ascending = False).values, color = ['red','grey'] )
plt.title('Arrival_time')
plt.xlabel('time of Day')
plt.ylabel('number of flights')
```

```python
plt.figure(figsize = (20,6))

plt.subplot(1,2,1)

plt.bar(df.source_city.value_counts().index, df.source_city.value_counts(ascending = False).values, color = ['red','grey'] )
plt.title('Flight Sources')
plt.xlabel('source_city')
plt.ylabel('number of flights')

plt.subplot(1,2,2)

plt.barh(df.destination_city.value_counts().index, df.destination_city.value_counts(ascending = False).values, color = ['green','yellow'] )
plt.title('Flight Destination')
plt.ylabel('destination_city')
plt.xlabel('number of flights')
```
