```python
deptCounts = df.groupby('Department')['Status'].agg(total_emp = 'count', resigned = lambda x : (x == 'Resigned').sum())
deptCounts
```

```python
deptCounts['attrition_Rate%'] = (deptCounts['resigned'] / deptCounts['total_emp'])*100
deptCounts
```

```python
deptCounts.sort_values('attrition_Rate%', ascending = False)
```
