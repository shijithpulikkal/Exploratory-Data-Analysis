```python
df = data.groupby('Album_type')[['Views', 'Likes', 'Comments']].mean()
df
```

```python
type(df)
```

```python
df = df.reset_index()
df
```

```python
df_melted = pd.melt(df, id_vars = 'Album_type', var_name = 'Segment', value_name = 'Values')
df_melted
```

```python
plt.figure(figsize = (9,4))

sns.barplot(x='Album_type', y='Values', hue = 'Segment', data = df_melted )
```
